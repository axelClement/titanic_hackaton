# Titanic Dataset - Exploratory Data Analysis

This notebook explores the Titanic dataset to understand the data and identify patterns for prediction.

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

# Set style
sns.set_style('whitegrid')
plt.rcParams['figure.figsize'] = (12, 6)

%matplotlib inline
```

## Load Data

```python
# Load training and test data
train_df = pd.read_csv('../data/train.csv')
test_df = pd.read_csv('../data/test.csv')

print(f"Training set shape: {train_df.shape}")
print(f"Test set shape: {test_df.shape}")
```

## Data Overview

```python
# First few rows
train_df.head()
```

```python
# Data types and missing values
train_df.info()
```

```python
# Statistical summary
train_df.describe()
```

```python
# Missing values
missing = pd.DataFrame({
    'Column': train_df.columns,
    'Missing Count': train_df.isnull().sum(),
    'Percentage': (train_df.isnull().sum() / len(train_df)) * 100
}).sort_values('Missing Count', ascending=False)

print(missing[missing['Missing Count'] > 0])
```

## Survival Analysis

```python
# Overall survival rate
survival_rate = train_df['Survived'].mean()
print(f"Overall survival rate: {survival_rate:.2%}")

# Survival counts
plt.figure(figsize=(8, 6))
train_df['Survived'].value_counts().plot(kind='bar')
plt.title('Survival Distribution')
plt.xlabel('Survived (0 = No, 1 = Yes)')
plt.ylabel('Count')
plt.xticks(rotation=0)
plt.show()
```

## Feature Analysis

```python
# Survival by Sex
plt.figure(figsize=(10, 5))
sns.barplot(data=train_df, x='Sex', y='Survived')
plt.title('Survival Rate by Sex')
plt.ylabel('Survival Rate')
plt.show()

print(train_df.groupby('Sex')['Survived'].agg(['mean', 'count']))
```

```python
# Survival by Passenger Class
plt.figure(figsize=(10, 5))
sns.barplot(data=train_df, x='Pclass', y='Survived')
plt.title('Survival Rate by Passenger Class')
plt.ylabel('Survival Rate')
plt.show()

print(train_df.groupby('Pclass')['Survived'].agg(['mean', 'count']))
```

```python
# Age distribution
plt.figure(figsize=(12, 5))

plt.subplot(1, 2, 1)
train_df['Age'].hist(bins=30)
plt.title('Age Distribution')
plt.xlabel('Age')
plt.ylabel('Count')

plt.subplot(1, 2, 2)
train_df[train_df['Survived']==1]['Age'].hist(bins=30, alpha=0.5, label='Survived')
train_df[train_df['Survived']==0]['Age'].hist(bins=30, alpha=0.5, label='Did not survive')
plt.title('Age Distribution by Survival')
plt.xlabel('Age')
plt.ylabel('Count')
plt.legend()

plt.tight_layout()
plt.show()
```

```python
# Fare distribution
plt.figure(figsize=(12, 5))

plt.subplot(1, 2, 1)
train_df['Fare'].hist(bins=50)
plt.title('Fare Distribution')
plt.xlabel('Fare')
plt.ylabel('Count')

plt.subplot(1, 2, 2)
train_df.boxplot(column='Fare', by='Survived')
plt.title('Fare by Survival')
plt.suptitle('')

plt.tight_layout()
plt.show()
```

```python
# Family size analysis
train_df['FamilySize'] = train_df['SibSp'] + train_df['Parch'] + 1

plt.figure(figsize=(10, 5))
sns.barplot(data=train_df, x='FamilySize', y='Survived')
plt.title('Survival Rate by Family Size')
plt.ylabel('Survival Rate')
plt.show()

print(train_df.groupby('FamilySize')['Survived'].agg(['mean', 'count']))
```

```python
# Embarked analysis
plt.figure(figsize=(10, 5))
sns.barplot(data=train_df, x='Embarked', y='Survived')
plt.title('Survival Rate by Embarkation Port')
plt.ylabel('Survival Rate')
plt.show()

print(train_df.groupby('Embarked')['Survived'].agg(['mean', 'count']))
```

## Correlation Analysis

```python
# Correlation matrix for numerical features
numerical_features = ['Survived', 'Pclass', 'Age', 'SibSp', 'Parch', 'Fare', 'FamilySize']
corr_matrix = train_df[numerical_features].corr()

plt.figure(figsize=(10, 8))
sns.heatmap(corr_matrix, annot=True, cmap='coolwarm', center=0, fmt='.2f')
plt.title('Correlation Matrix')
plt.show()
```

## Key Findings

Based on the exploratory analysis:

1. **Sex**: Females had a much higher survival rate than males
2. **Pclass**: First class passengers had higher survival rates
3. **Age**: Children had higher survival rates
4. **Family Size**: Passengers with small families (2-4 members) had better survival rates
5. **Fare**: Higher fare correlated with higher survival rates
6. **Embarked**: Port of embarkation showed some correlation with survival

These insights will guide feature engineering and model development.
